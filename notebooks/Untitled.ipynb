```python
A = torch.randn(10, 3, 20)
B = torch.randn(10, 3, 20)
C = torch.nn.Parameter(torch.randn(20,3))

# Perform batched matrix multiplication
# Output C shape will be: [Batch, M, N]
D = torch.einsum('bij, bjk -> bik', torch.cat((A,B),dim=1), C.unsqueeze(0))

print(D.shape)  # torch.Size([10, 3, 5])
```
